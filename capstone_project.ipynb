```python
resi_permits_df = permits_df[permits_df['Permit_type_description'].str.contains('Residential', case=False, na=False)
    | permits_df['Permit_subtype_description'].str.contains('Multifamily', case=False, na=False)]
resi_permits_df.head(60)
```

```python
com_permit_df = permits_df[permits_df['Permit_type_description'].str.contains('Commercial', case=False, na=False)]
com_permit_df.head(60)
```

```python
res_permits_df = permits_df[permits_df['Permit_type_description'].str.contains('Residential', case=False, na=False)]
res_permits_df.head(50)
```

```python
resi_single_fam_df = res_permits_df[res_permits_df['Permit_subtype_description'] == 'Single Family Residence']
resi_single_fam_df
```

```python
multi_fam_build_df = com_permit_df[com_permit_df['Permit_subtype_description'].str.contains('Multifamily', case=False, na=False)]
multi_fam_build_df.head(50)
```

```python
commercial_only_df = com_permit_df[
    (com_permit_df['Permit_type_description'].str.contains('Commercial', case=False, na=False)) & 
    (~com_permit_df['Permit_subtype_description'].str.contains('Multifamily', case=False, na=False))
]
commercial_only_df.head(60)
```

```python
violation_df.head(60)
```

```python
violation_grass_df = violation_df[violation_df['violations_noted'].str.contains('HIGH WEEDS', case=False, na=False)]
violation_grass_df
```

```python
violation_junk_df = violation_df[violation_df['violations_noted'].str.contains('Junk' or 'Trash', case=False, na=False)]
violation_junk_df
```

```python
violation_df.head(60)
```

```python

violation_resolved_df = violation_df[violation_df['last_act__result'].str.contains('Violation Resolved', case=False, na=False)]
violation_resolved_df
```

```python

violation_open_df = violation_df[violation_df['status'].str.contains('OPEN', case=False, na=False)]
violation_open_df.head(60)
```

```python
violation_count_by_zip = (
    violation_df
      .groupby('zipcode')
      .size()
      .reset_index(name='total_violation_requests')
      .sort_values(by='total_violation_requests', ascending=False)
)
print(violation_count_by_zip)
```

```python
open_count_by_zip = (
    violation_open_df
    .groupby('zipcode')
    .size()
    .reset_index(name='open_violation_requests')
    .sort_values(by='open_violation_requests', ascending=False)
)

print(open_count_by_zip)
```

```python
resolved_count_by_zip = (
    violation_resolved_df
    .groupby('zipcode')
    .size()
    .reset_index(name='resolved_violation_requests')
    .sort_values(by='resolved_violation_requests', ascending=False)
)

print(resolved_count_by_zip)
```

```python
violations_by_zip_type = (
    violation_df
    .groupby(['zipcode', 'violations_noted'])
    .size()
    .reset_index(name='total_requests')
    .sort_values(by='total_requests', ascending=False)
)

violations_by_zip_type.head(60)
```

```python
violation_resolved_df['date_received'] = pd.to_datetime(violation_resolved_df['date_received'])
```

```python

violation_resolved_df['last_activity_date'] = pd.to_datetime(violation_resolved_df['last_activity_date'])
```

```python
violation_resolved_df['days_to_resolve'] = (
    violation_resolved_df['last_activity_date'] - violation_resolved_df['date_received']
).dt.days
violation_resolved_df.head(50)
```

```python
avg_resolution_by_zip = (
    violation_resolved_df
    .groupby('zipcode')['days_to_resolve']
    .mean()
    .reset_index(name='avg_days_to_resolve').round(1)
    .sort_values(by='avg_days_to_resolve', ascending=False)
)

print(avg_resolution_by_zip)
```

```python
total_permit_count_by_zip = (
    permits_df
    .groupby('Zipcode')
    .size()
    .reset_index(name='total_permits')
    .sort_values(by='total_permits', ascending=False)
)

print(total_permit_count_by_zip)
```

```python
resi_permit_count_by_zip = (
    resi_permits_df
    .groupby('Zipcode')
    .size()
    .reset_index(name='total_resi_permits')
    .sort_values(by='total_resi_permits', ascending=False)
)

print(resi_permit_count_by_zip)
```

```python
com_permit_count_by_zip = (
    commercial_only_df
    .groupby('Zipcode')
    .size()
    .reset_index(name='total_com_permits')
    .sort_values(by='total_com_permits', ascending=False)
)

print(com_permit_count_by_zip)
```

```python
permits_df.columns
```

```python
const_cost_by_zip = (permits_df
 .groupby('Zipcode')['Construction_cost']
 .sum()
 .reset_index(name='total_construction_cost')
 .sort_values(by='total_construction_cost', ascending=False)
)

const_cost_by_zip.head(60).style.format({'total_construction_cost': '${:,}'})
```

```python
const_cost_by_zip = const_cost_by_zip.rename(columns={'Zipcode': 'zipcode'})
const_cost_by_zip.head(1)
```

```python
## To see how much is being invested per zip and how many violation requests have been submitted. To see if there's a correlation. 

const_cost_and_violation_count_by_zip = pd.merge(
    const_cost_by_zip, 
    violation_count_by_zip, 
    on='zipcode', 
    how='outer',
) .sort_values(by='total_construction_cost', ascending=False)
const_cost_and_violation_count_by_zip.head(60).style.format({'total_construction_cost': '${:,}'})
```

```python
const_cost_and_violation_count_by_zip = const_cost_and_violation_count_by_zip.fillna(0)
const_cost_and_violation_count_by_zip.head(60)
```

```python
total_permit_count_by_zip 
```

```python
total_permit_count_by_zip = total_permit_count_by_zip.rename(columns={'Zipcode': 'zipcode'})
```

```python
permits_cost_and_violations_by_zip = pd.merge(
    total_permit_count_by_zip,
    const_cost_and_violation_count_by_zip,
    on='zipcode',
    how='outer'
).fillna(0).sort_values(by='total_permits', ascending=False)
permits_cost_and_violations_by_zip
```

```python
permits_cost_and_violations_by_zip['violations_per_permit'] = (
    permits_cost_and_violations_by_zip['total_violation_requests'] / 
    permits_cost_and_violations_by_zip['total_permits']
).fillna(0)

permits_cost_and_violations_by_zip
```

```python
permits_cost_and_violations_by_zip['avg_cost_per_permit'] = (
    permits_cost_and_violations_by_zip['total_construction_cost'] / 
    permits_cost_and_violations_by_zip['total_permits']
).fillna(0)
permits_cost_and_violations_by_zip
```

```python
permits_cost_and_violations_by_zip['valuation_millions'] = (
    permits_cost_and_violations_by_zip['total_construction_cost'] / 1_000_000
).round(2)

permits_cost_and_violations_by_zip
```

```python
import plotly.express as px

Construction_Valuation_vs_Violations = px.scatter(
    permits_cost_and_violations_by_zip,
    x='valuation_millions',
    y='total_violation_requests',
    hover_name='zipcode',
    title='Construction Valuation ($M) vs. Property Violations',
    labels={
        'valuation_millions': 'Total Construction Valuation ($ Millions)',
        'total_violation_requests': 'Total Property Violations'
    },
    width=1100,   # Adjust width in pixels
    height=1000    # Adjust height in pixels
)

Construction_Valuation_vs_Violations.show()
```

```python
median_valuation_costs = permits_cost_and_violations_by_zip['total_construction_cost'].median()
median_valuation_costs
```

```python
median_valuation = permits_cost_and_violations_by_zip['valuation_millions'].median()
median_valuation
```

```python
median_violations = permits_cost_and_violations_by_zip['total_violation_requests'].median()
median_violations
```

```python
stable_growth_zips = permits_cost_and_violations_by_zip[
    (permits_cost_and_violations_by_zip['valuation_millions'] > median_valuation) & 
    (permits_cost_and_violations_by_zip['total_violation_requests'] < median_violations)
].sort_values(by='valuation_millions', ascending=False)

stable_growth_zips[['zipcode', 'valuation_millions', 'total_violation_requests', 'total_permits']]
```

```python
redevelopment_zips = permits_cost_and_violations_by_zip[
    (permits_cost_and_violations_by_zip['valuation_millions'] > median_valuation) & 
    (permits_cost_and_violations_by_zip['total_violation_requests'] > median_violations)
].sort_values(by='valuation_millions', ascending=False)

redevelopment_zips[['zipcode', 'valuation_millions', 'total_violation_requests', 'total_permits']]
```
