```python
resi_permits_df = permits_df[permits_df['Permit_type_description'].str.contains('Residential', case=False, na=False)]
resi_permits_df.head(10)
```

```python
com_permit_df = permits_df[permits_df['Permit_type_description'].str.contains('Commercial', case=False, na=False)]
com_permit_df.head(60)
```

```python
res_permits_df = permits_df[permits_df['Permit_type_description'].str.contains('Residential', case=False, na=False)]
res_permits_df.head(50)
```

```python
resi_single_fam_df = res_permits_df[res_permits_df['Permit_subtype_description'] == 'Single Family Residence']
resi_single_fam_df
```

```python
multi_fam_build_df = com_permit_df[com_permit_df['Permit_subtype_description'].str.contains('Multifamily', case=False, na=False)]
multi_fam_build_df.head(50)
```

```python
commercial_only_df = com_permit_df[
    (com_permit_df['Permit_type_description'].str.contains('Commercial', case=False, na=False)) & 
    (~com_permit_df['Permit_subtype_description'].str.contains('Multifamily', case=False, na=False))
]
commercial_only_df.head(60)
```

```python
violation_df.head(60)
```

```python
violation_grass_df = violation_df[violation_df['violations_noted'].str.contains('HIGH WEEDS', case=False, na=False)]
violation_grass_df
```

```python
violation_junk_df = violation_df[violation_df['violations_noted'].str.contains('Junk' or 'Trash', case=False, na=False)]
violation_junk_df
```

```python
violation_df.head(60)
```

```python

violation_resolved_df = violation_df[violation_df['last_act__result'].str.contains('Violation Resolved', case=False, na=False)]
violation_resolved_df
```

```python

violation_open_df = violation_df[violation_df['status'].str.contains('OPEN', case=False, na=False)]
violation_open_df.head(60)
```

```python
violation_count_by_zip = (
    violation_df
      .groupby('zipcode')
      .size()
      .reset_index(name='total_violation_requests')
      .sort_values(by='total_violation_requests', ascending=False)
)
print(violation_count_by_zip)
```
